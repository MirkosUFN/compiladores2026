```python
import sys
import os

def remove_c_comments(code: str) -> str:
    """
    Remove comentários de estilo C (// e /* ... */) preservando a estrutura
    de linhas e colunas do código fonte original.
    """
    result = []
    i = 0
    n = len(code)

    # Estados possíveis: 'NORMAL', 'IN_STRING', 'IN_CHAR', 'IN_LINE_COMMENT', 'IN_BLOCK_COMMENT'
    state = 'NORMAL'

    while i < n:
        char = code[i]
        next_char = code[i + 1] if i + 1 < n else ''

        if state == 'NORMAL':
            if char == '/' and next_char == '/':
                state = 'IN_LINE_COMMENT'
                result.append('  ')  # Substitui '//' por 2 espaços
                i += 2
            elif char == '/' and next_char == '*':
                state = 'IN_BLOCK_COMMENT'
                result.append('  ')  # Substitui '/*' por 2 espaços
                i += 2
            elif char == '"':
                state = 'IN_STRING'
                result.append(char)
                i += 1
            elif char == "'":
                state = 'IN_CHAR'
                result.append(char)
                i += 1
            else:
                result.append(char)
                i += 1

        elif state == 'IN_STRING':
            result.append(char)
            # Trata caractere de escape (ex: \")
            if char == '\\' and i + 1 < n:
                result.append(code[i + 1])
                i += 2
            elif char == '"':
                state = 'NORMAL'
                i += 1
            else:
                i += 1

        elif state == 'IN_CHAR':
            result.append(char)
            # Trata caractere de escape (ex: \')
            if char == '\\' and i + 1 < n:
                result.append(code[i + 1])
                i += 2
            elif char == "'":
                state = 'NORMAL'
                i += 1
            else:
                i += 1

        elif state == 'IN_LINE_COMMENT':
            if char == '\n':
                state = 'NORMAL'
                result.append('\n')  # Preserva a quebra de linha
            elif char == '\\' and next_char == '\n':
                # Continuação de linha em comentário C (\ no final da linha)
                result.append(' \n')
                i += 2
                continue
            else:
                result.append(' ')   # Substitui caractere por espaço
            i += 1

        elif state == 'IN_BLOCK_COMMENT':
            if char == '*' and next_char == '/':
                state = 'NORMAL'
                result.append('  ')  # Substitui '*/' por 2 espaços
                i += 2
            elif char == '\n':
                result.append('\n')  # Preserva a quebra de linha
                i += 1
            else:
                result.append(' ')   # Substitui caractere por espaço
                i += 1

    return "".join(result)


def main():
    if len(sys.argv) < 2:
        print("Uso: python remove_comments.py <arquivo.c> [arquivo_saida.c]")
        sys.exit(1)

    input_filepath = sys.argv[1]

    if not os.path.isfile(input_filepath):
        print(f"Erro: O arquivo '{input_filepath}' não foi encontrado.")
        sys.exit(1)

    # Determina o arquivo de saída
    if len(sys.argv) >= 3:
        output_filepath = sys.argv[2]
    else:
        name, ext = os.path.splitext(input_filepath)
        output_filepath = f"{name}_no_comments{ext}"

    with open(input_filepath, 'r', encoding='utf-8', errors='ignore') as f:
        c_code = f.read()

    cleaned_code = remove_c_comments(c_code)

    with open(output_filepath, 'w', encoding='utf-8') as f:
        f.write(cleaned_code)

    print(f"Comentários removidos com sucesso!")
    print(f"Arquivo gerado: {output_filepath}")


if __name__ == "__main__":
    main()
```
