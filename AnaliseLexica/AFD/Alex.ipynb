```python
caminho = '/content/drive/MyDrive/2026/2 semestre/compiladores/configAFD.txt'

estados = []
estadosFinais = []
simbolos = []
regrasTransicao = []

#abrindo o arquivo de configuração do AFD
config = open(caminho)
#lendo a primeira linha - estados
estados = config.readline().strip().split(' ')
estadoInicial = estados[0]
#lendo a segunda linha - símbolos
simbolos = config.readline().strip().split(' ')
#lendo a terceira linha - estadosFinais
estadosFinais = config.readline().strip().split(' ')
#lendo a partir da quarta linha - regras transição
linha = config.readline()
while(linha):
  regras = linha.strip().split(' ')
  for r in regras:
    regrasTransicao.append(r)
  linha = config.readline() #lê a próxima linha

config.close()#fecha o arquivo de configuração

def reconheceTermo(termo):
  estadoAtual = estadoInicial #inicio o AFD toda vez que for reconhecer um termo
  numCaracteres = len(termo) #número de caracteres do termo
  for i in range(numCaracteres):
    if termo[i] not in simbolos:
      print('Símbolo não reconhecido')
      return
    #devemos andar no AFD
    padrao = estadoAtual+':'+termo[i]
    for i in range(len(regrasTransicao)):
      regra = regrasTransicao[i]
      if regra.startswith(padrao):
        reg = regra.split(':')#estadoAtual:simbolo:estadoFinal
        estadoAtual = reg[2]
    #agora vamos procurar o estado atual nos estados finais
  for f in range(len(estadosFinais)):
    if estadosFinais[f].startswith(estadoAtual):
      return termo+':'+estadosFinais[f].split(':')[1]
  #print(estadoAtual)


arquivo = open('/content/drive/MyDrive/2026/2 semestre/compiladores/numeros.txt')
print(reconheceTermo(arquivo.readline().strip()))
print(reconheceTermo(arquivo.readline().strip()))
print(reconheceTermo(arquivo.readline().strip()))
```
